Reproduction du pipeline complet.


```python
import sys
sys.path.insert(0, '../backend')
from automl.pipeline import AutoMLPipeline
from automl.agents.qa_agent import QAAgent
import pandas as pd
```

```python
pipeline = AutoMLPipeline(source='../data/samples/titanic.csv', target='Survived', optuna_trials=5, cv_folds=3)
result = pipeline.run()
result.leaderboard[0].name
```

```python
pd.DataFrame([{'model': m.name, 'metric': m.metrics, 'time_s': m.training_time_s} for m in result.leaderboard])
```

```python
qa = QAAgent(pd.read_csv('../data/samples/titanic.csv'), target='Survived', leaderboard=result.leaderboard)
qa.ask('Pourquoi la cible est-elle desequilibree ?')
```
